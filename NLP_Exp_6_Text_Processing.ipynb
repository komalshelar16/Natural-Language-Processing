```python
!pip install -q spacy
!python -m spacy download en_core_web_sm
import spacy
import nltk
from nltk.stem import PorterStemmer
nlp = spacy.load("en_core_web_sm")
stemmer = PorterStemmer()
```

```python
text = "Appleeee Inc. is an American multinational technology company headquartered in Cupertino, California, in Silicon Valley, best known for its consumer electronics, software and online services. Founded in 1976 as Apple Computer Company by Steve Jobs, Steve Wozniak and Ronald Wayne, the company was incorporated by Jobs and Wozniak as Apple Computer, Inc. the following year. It was renamed to its current name in 2007 as the company had expanded its focus from computers to consumer electronics. Apple is one of the Big Tech companies."
print("Original Text:", text)
```

```python
doc = nlp(text)
tokens = [token.text for token in doc]

print("Original text:", text)
print("Tokens:", tokens)
```

```python
for token in doc[:10]:
    print(f"{token.text:10} -> {stemmer.stem(token.text)}")
```

```python
for token in doc[:10]:
    print(f"{token.text:10} -> {token.lemma_}")
```

```python
for token in doc[:10]:
    print(f"{token.text:10} -> {token.pos_}")
```

```python
for ent in doc.ents:
    print(f"{ent.text:20} -> {ent.label_}")
```

```python
filtered = [token.text for token in doc if not token.is_stop]
print("Without stop words:", " ".join(filtered))
```

```python
for word in tokens:
    print(f"{word:15} → {word}")
```

```python
def remove_consecutive_repeated_chars(word):
    if not word:
        return ""
    result = [word[0]]
    for char in word[1:]:
        if char != result[-1]:
            result.append(char)
    return "".join(result)
```

```python
for word in tokens:
    print(f"{word:15} → {remove_consecutive_repeated_chars(word)}")
```

```python
import re

text = "Helloooo!!!"
result = re.sub(r'(.)\1{2,}', r'\1\1', text)
print(result)
```

```python
def remove_repeated_chars(text):
    result = text[0]
    for char in text[1:]:
        if char != result[-1]:
            result += char
    return result
print(remove_repeated_chars("Helloooo!!!"))
```

```python
!pip install pyspellchecker
from spellchecker import SpellChecker
import re

quote = """Ohhhhh sorry. I am runnnnnning late for classsssssss!!!!!!!! I didn't bump into on purposeeee! """
sample_text = nlp(quote)
sentences = [sentence.text.lower() for sentence in sample_text.sents]
change = ""
for sentence in sentences:
  change += " " + re.sub(r'(.)\1+', r'\1', sentence)
change = change.strip()
sample_text = nlp(change)
spell_checker = SpellChecker()
corrected_words = []
for word in sample_text:
  if word.is_alpha:
    correct = spell_checker.correction(word.text)
    corrected_words.append(correct)
  else:
    corrected_words.append(word.text)
corrected_sentences = " "
for word in corrected_words:
  if word in r"\b[.!?,] | n't | 'll\b":
    corrected_sentences += word
  else:
    corrected_sentences += " " + word
corrected_sentences = corrected_sentences.strip()
sample_text = nlp(corrected_sentences)
print(sample_text)
```
